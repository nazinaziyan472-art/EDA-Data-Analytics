```python
sns.set_style("whitegrid")
plt.Figure(figsize=(12,7))
sns.histplot(data=df,x='Age',bins=20,kde=True,color='green')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()
```

```python
sns.boxplot(data=df,y='Age',color='purple')
plt.title("Before Removing Outliers - Age")
plt.show()
```

```python
q1=np.percentile(df['Age'],25)
q3=np.percentile(df['Age'],75)
iqr = q3 -q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

clean_df = df[(df['Age'] >= lower) & (df['Age']<= upper)]
```

```python
sns.boxplot(y='Age',data=clean_df,color='purple')
plt.title("After Removing Outliers - Age")
plt.show()

```

```python
sns.countplot(data=df,x='Geography',hue='Geography',palette={'France':'red','Spain':'green','Germany':'yellow'},legend=True)
plt.title("Number of customers by Geography")
plt.xlabel("country")
plt.ylabel("number of customers")
plt.show()
```

```python
sns.countplot(data=df,x='Gender',hue='Gender',palette={'Female':'pink','Male':'Blue'},legend=True)
plt.title("Gender Distribution of customers")
plt.xlabel("Gender")
plt.ylabel("Number of customers")
plt.show()
```

```python
sns.countplot(data=df,x='Exited',hue='Exited',palette={0:'purple',1:'green'})
plt.title("Number of customers Exited")
plt.xlabel("Exited")
plt.ylabel("Number of customers")
plt.legend(labels=["Not Exited","Exited"],title="customer status")
plt.show()
```

```python
sns.boxplot(data=df,x='Exited',y='Age',hue='Exited',palette={0:'yellow',1:"orange"})
plt.title("before Removing Outliers")
plt.legend(title='Customer Status',loc='upper right',labels=['Not Exited','Exited'])
plt.show()
```

```python
q1=df.groupby('Exited')['Age'].transform(lambda x:x.quantile(0.25))
q3=df.groupby('Exited')['Age'].transform(lambda x:x.quantile(0.75))
iqr=q3-q1
lower=q1-1.5*iqr
upper=q3-1.5*iqr
clean_df=df[(df['Age']>=lower)&(df['Age']<=upper)]
```

```python
clean_df
```

```python
sns.boxplot(data=clean_df,x='Exited',y='Age',hue='Exited',palette={0:'yellow',1:"orange"})
plt.title('After Removing Outliers')
plt.xlabel('Exited')
plt.ylabel('Age')
plt.legend(title='Customer Status',loc='upper right',labels=['Not Exited','Exited'])
plt.show()
```

```python
sns.countplot(data=df,x='Geography',hue='Exited',palette={0:'green',1:'red'})
plt.title("Customer churn by Geography")
plt.xlabel("Geography")
plt.ylabel("Number of Customers")
plt.legend(title="Customer Status",labels=["Not Exited","Exited"])
plt.show()
```
